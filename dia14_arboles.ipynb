```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Dataset de ejemplo (ajústalo si usas otro)
data = pd.DataFrame({
    "Agua": [300, 280, 310, 290, 305, 315, 295],
    "Gasolina": [150, 160, 170, 180, 175, 165, 155],
    "Luz": [200, 210, 190, 205, 195, 215, 185],
    "Renta": [5000, 5000, 5000, 5000, 5000, 5000, 5000],
    "Super": [1200, 1300, 1100, 1250, 1280, 1180, 1220],
    "Total": [6850, 6980, 6770, 6925, 6955, 7075, 6855]
})

# Separar variables (X) y target (y)
X = data.drop("Total", axis=1)
y = data["Total"]

# División train/test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python

from sklearn.tree import DecisionTreeRegressor

# Crear el árbol
tree = DecisionTreeRegressor(max_depth=3, random_state=42)

# Entrenar
tree.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

# Predicciones
y_pred_tree = tree.predict(X_test)

# Evaluación
tree_rmse = np.sqrt(mean_squared_error(y_test, y_pred_tree))
print("Decision Tree RMSE:", tree_rmse)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(12,6))
plot_tree(tree, feature_names=X.columns, filled=True, rounded=True, fontsize=10)
plt.show()

```
